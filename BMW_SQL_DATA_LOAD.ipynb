```python
import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import quote_plus

# Password encoded
password = quote_plus("Devesh@123")

# Connection
engine = create_engine(f'mysql+pymysql://root:{password}@localhost/bmw_sales')

# Test connection first
try:
    with engine.connect() as conn:
        print("✅ Connected to MySQL successfully!")
except Exception as e:
    print(f"❌ Connection failed: {e}")
    exit()

# Read CSV
print("Reading CSV...")
df = pd.read_csv(r'C:\Users\Dell\BMW_Sales_Cleaned_v1.csv')
print(f"Rows: {len(df)}, Columns: {len(df.columns)}")

# Import to MySQL
print("Importing to MySQL...")
df.to_sql(
    name='bmw_sales_data',
    con=engine,
    if_exists='replace',
    index=False,
    chunksize=500
)

print(f"✅ Done! {len(df)} rows imported successfully!")
```
